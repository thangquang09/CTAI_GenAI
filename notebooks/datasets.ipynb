```python
from datasets import load_dataset

# Login using e.g. `huggingface-cli login` to access this dataset
ds = load_dataset("Wix/WixQA", "wixqa_expertwritten", split="train")
corpus = load_dataset("Wix/WixQA", "wix_kb_corpus", split="train")
simulated = load_dataset("Wix/WixQA", "wixqa_simulated", split="train")
```

```python
for d in ds:
    articles = d['article_ids']
    if len(articles) > 1:
        print(d)
        break
```

```python
num_characters = []
num_words = []

for d in corpus:
    contents = d['contents']
    num_characters.append(len(contents))
    num_words.append(len(contents.split()))
```

```python
import matplotlib.pyplot as plt

plt.hist(num_characters, bins=50)
plt.title("Number of character")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.hist(num_words, bins=50)
plt.title("Number of words")
plt.show()
```
